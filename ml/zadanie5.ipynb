$ x \sim N(\mu, \Sigma) $

$\mu$ -> mean vector

$\Sigma$ -> covariance matrix

$f(x,y) = \frac{1}{2\pi|\Sigma|^{\frac{1}{2}}} exp(-\frac{1}{2}(x-\mu)^T\Sigma^{-1}(x-\mu)) $

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

mean = [0,0]
cov = [[1,0],[0,1]]

x, y = np.random.randn(2,10**4)
```

```python
# import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
hist, xe, ye = np.histogram2d(x, y, bins=40, range=[[-3, 3], [-3, 3]], density=True)

xc = (xe[:-1] + xe[1:]) / 2
yc = (ye[:-1] + ye[1:]) / 2
X, Y = np.meshgrid(xc, yc)

Z = (1 / (2 * np.pi)) * np.exp(-0.5 * (X**2 + Y**2))

fig = go.Figure()

fig.add_trace(go.Surface(x=X,y=Y,z=hist.T,colorscale='Viridis',opacity=0.8))
fig.add_trace(go.Surface(x=X,y=Y,z=Z,colorscale='Reds',opacity=0.6))

fig.show()
```
