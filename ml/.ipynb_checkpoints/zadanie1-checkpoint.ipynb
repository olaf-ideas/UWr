```python
from ucimlrepo import fetch_ucirepo

iris = fetch_ucirepo(id=53)

X = iris.data.features
y = iris.data.targets
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,8))
plt.scatter(X.index, X['sepal length'])
plt.grid(True)
plt.show()

print(X['sepal length'].min(), X['sepal length'].max())
```

```python
plt.figure(figsize=(8,8))
X_setosa = X[y['class'] == 'Iris-setosa']


x_setosa = []
y_setosa = []
last = 0
now = 2.5
while now < 10:
    cnt = len(X_setosa[X_setosa['sepal width'].between(last, now)])
    if cnt > 0:
        x_setosa.append(f'[{last}, {now}]')
        y_setosa.append(cnt)
        last = now
    now += 0.5

plt.bar(x_setosa, y_setosa)
plt.show()
```

```python
plt.figure(figsize=(8,8))

#plt.scatter(X['sepal length'], X['sepal width'])
#plt.show()

print(y['class'].unique())
colors = {
    'Iris-setosa' : 'red',
    'Iris-versicolor' : 'green',
    'Iris-virginica': 'blue'
}

for species, color in colors.items():
    iris_species = X[y['class'] == species]
    plt.scatter(iris_species['sepal length'], iris_species['sepal width'], color=colors[species])
plt.show()
```
