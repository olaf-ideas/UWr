# Zadanie 4

```ocaml
let parens_ok s =
    let rec help s acc =
        match s with
        | [] -> (acc = 0)
        | '(' :: s' -> help s' (acc + 1)
        | ')' :: s' -> if acc = 0 then false else help s' (acc - 1)
        | _ :: s' -> false
    in
    help (List.of_seq (String.to_seq s)) 0
```

```ocaml
parens_ok "(())()";;
parens_ok "()())";;
parens_ok "((())";;
parens_ok "(x)";;
```

# Zadanie 5

```ocaml
let parens_ok s =
    let rec help s stk =
        match s, stk with
        | [], _ -> stk = []
        | '(' :: s', _ -> help s' ('(' :: stk)
        | '[' :: s', _ -> help s' ('[' :: stk)
        | '{' :: s', _ -> help s' ('{' :: stk)
        | ')' :: s', '(' :: stk'
        | ']' :: s', '[' :: stk'
        | '}' :: s', '{' :: stk' -> help s' stk'
        | _, _ -> false
    in
    help (List.of_seq (String.to_seq s)) []
```

```ocaml
 parens_ok "()[]";;
 parens_ok "[()[{}]]";;
 parens_ok "{(]}";;
 parens_ok "({)";;
```

```ocaml
type 'a symbol = 
    | T of string
    | N of 'a

type 'a grammar = ('a * ('a symbol list) list) list

let rec generate g s =
    let lista_rozw = (List.assoc s g) in
    let rand = Random.int (List.length lista_rozw) in
    let rec help rozw acc =
        match rozw with
        | [] -> acc
        | x :: rozw' ->
            match x with
            | T (nxt) -> help rozw' (String.concat "" [acc;nxt])
            | N (nxt) -> help rozw' (String.concat "" [acc;generate g nxt])
    in help (List.nth lista_rozw rand) ""

let expr : unit grammar =
    [() , [[ N (); T " + " ; N ()];
        [ N (); T " * " ; N ()];
        [ T " ( " ; N (); T " ) " ];
        [ T " 1 " ];
        [ T " 2 " ]]]
```

```ocaml
generate expr ();;
(* - : string = " (((1*1*2))) "*)
generate expr ();;
(* - : string = " 2 " *)
generate expr ();;
(* - : string = " 1+1*1 " *)
generate expr ();;
```
