```ocaml
(* Olaf Surgut, Version 1.1.0 *)

let ( let* ) xs ys = List.concat_map ys xs

let rec choose m n =
  if m > n then [] else m :: choose (m+1) n

(* somehow make it iterative ??? *)
let rec build_row ps n =
    match ps, n with
    | [], n -> [List.init n (fun y -> false)]
    | _, 0 -> []
    | x :: [], _ ->
        if x > n then
            []
        else
            let take = build_row [] (n - x) in
            let dont = build_row ps (n - 1) in
            (List.map (fun l -> (List.init x (fun y -> true)) @ l) take) @
            (List.map (fun x -> false :: x) dont)
    | x :: ps', _ ->
        if x >= n then
            []
        else
            let take = build_row ps' (n - x - 1) in
            let dont = build_row ps  (n - 1) in
            (List.map (fun l -> (List.init x (fun y -> true)) @ (false :: l)) take) @
            (List.map (fun x -> false :: x) dont)

let rec build_candidate pss n =
    match pss with
    | [] ->
        failwith "pss cannot be empty"
    | ps :: [] ->
        List.map (fun x -> [x]) (build_row ps n)
    | ps :: pss' ->
        let* candidate = build_row ps n in
        let* rest = build_candidate pss' n in
            [candidate :: rest]

let verify_row ps xs = 
    let rec get_ps ys res =
        match ys with
        | [] -> res
        | y :: ys' ->
            match res with
            | [] -> 
                if y = true then
                    get_ps ys' [1]
                else
                    get_ps ys' [0]
            | r :: res' ->
                if y = true then
                    get_ps ys' ((r + 1) :: res')
                else
                if r = 0 then
                    get_ps ys' (0 :: res')
                else
                    get_ps ys' (0 :: r :: res')
    in
    match get_ps xs [] with
    | [] -> [] = ps
    | q :: qs ->
        if q = 0 then
            (List.rev qs) = ps
        else
            (List.rev (q :: qs)) = ps

let rec verify_rows pss xss =
    match pss, xss with
    | [], _ -> true
    | _, [] -> true
    | ps :: pss', xs :: xss' ->
        (verify_row ps xs) && (verify_rows pss' xss')
            
let transpose xss = 
    let rec help xss res = 
        match xss with
        | [] -> List.rev res
        | [] :: xss' -> help [] res
        | (x :: xs) :: xss' ->
            help (List.map List.tl xss) ((x :: (List.map List.hd xss')) :: res)
    in help xss []

type nonogram_spec = {rows: int list list; cols: int list list}

let solve_nonogram nono =
    build_candidate (nono.rows) (List.length (nono.cols))
    |> List.filter (fun xss -> transpose xss |> verify_rows nono.cols)

let example_1 = {
  rows = [[2];[1];[1]];
  cols = [[1;1];[2]]
}

let example_2 = {
  rows = [[2];[2;1];[1;1];[2]];
  cols = [[2];[2;1];[1;1];[2]]
}

let big_example = {
  rows = [[1;2];[2];[1];[1];[2];[2;4];[2;6];[8];[1;1];[2;2]];
  cols = [[2];[3];[1];[2;1];[5];[4];[1;4;1];[1;5];[2;2];[2;1]]
}
```

```ocaml
solve_nonogram example_1
```

```ocaml
solve_nonogram example_2
```
