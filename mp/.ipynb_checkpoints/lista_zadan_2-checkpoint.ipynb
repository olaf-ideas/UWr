```ocaml
let rec fib n = 
    match n with
    | 0 -> 0
    | 1 -> 1
    | _ -> fib (n - 1) + fib (n - 2)
```

```ocaml
let fib_iter n =
    let rec help a b n =
        match n with
        | 0 -> a
        | _ -> help b (a + b) (n - 1)
    in help 0 1 n
```

```ocaml
fib 0;;
fib 1;;
fib 2;;
fib 3;;
fib 4;;

fib_iter 0;;
fib_iter 1;;
fib_iter 2;;
fib_iter 3;;
fib_iter 4;;
fib_iter 80;;
```

```ocaml
let matrix_mult m n =
    let a, b, c, d = m and
        e, f, g, h = n
        in
    (a * e + b * g, a * f + b * h,
     c * e + d * g, c * f + d * h)
```

```ocaml
(*
    (a, b,
     c, d)
*)

let matrix_id = (1, 0, 0, 1)
```

```ocaml
let rec matrix_expt m k =
    match k with
    | 0 -> matrix_id
    | 1 -> m
    | _ -> matrix_mult m (matrix_expt m (k - 1))
```

```ocaml
let fib_matrix k =
    let _, fk, _, _ = matrix_expt (1, 1, 1, 0) k
    in fk
```

```ocaml
fib_matrix 0;;
fib_matrix 1;;
fib_matrix 2;;
fib_matrix 3;;
fib_matrix 4;;
fib_matrix 5;;
fib_matrix 6;;
fib_matrix 7;;
```

```ocaml
let rec matrix_expt_fast m k =
    match k with
    | 0 -> matrix_id
    | _ ->
        match k mod 2 with
        | 0 -> let m2 = matrix_expt_fast m (k / 2) in matrix_mult m2 m2
        | _ -> matrix_mult m (matrix_expt_fast m (k - 1))
```

```ocaml
let fib_fast k =
    let _, fk, _, _ = matrix_expt_fast (1, 1, 1, 0) k
    in fk
```

```ocaml
fib_fast 0;;
fib_fast 1;;
fib_fast 2;;
fib_fast 3;;
fib_fast 4;;
fib_fast 5;;
fib_fast 6;;
fib_fast 7;;
```

```ocaml
fib_matrix 80
```

```ocaml
fib_fast 80
```

```ocaml
let rec mem x xs =
    match xs with
    | [] -> false
    | y :: xs' ->
        if y = x then true else mem x xs'
```

```ocaml
mem 2 [1; 2; 3]
```

```ocaml
mem 4 [1; 2; 3]
```

```ocaml
let rec maximum xs =
    match xs with
    | [] -> neg_infinity
    | x :: xs' ->
        max (maximum xs') x
```

```ocaml
maximum [1.; 5.; 0.; 7.; 1.; 4.; 1.; 0.]
```

```ocaml
maximum []
```

```ocaml
let rec suffixes xs =
    match xs with
    | [] -> [[]]
    | x :: xs' ->
        xs :: suffixes xs'
```

```ocaml
suffixes [1; 2; 3; 4]
```

```ocaml
let rec is_sorted xs =
    match xs with
    | [] -> true
    | [x] -> true
    | x :: y :: xs' ->
        if x > y then false else is_sorted (y :: xs')
```

```ocaml
is_sorted [1; 3; 7; 7; 8;]
```

```ocaml
is_sorted [1; 3; 4; 5; 4;]
```

```ocaml
is_sorted [5; 1]
```

```ocaml
is_sorted [1]
```

```ocaml
let rec select xs =
    match xs with
    | [] -> failwith "empty list"
    | [x] -> (x, [])
    | x :: xs' ->
        let y, ys' = select xs' in
        if x < y then (x, y :: ys') else (y, x :: ys')
```

```ocaml
select [4; 3; 1; 2; 5]
```

```ocaml
let rec select_sort xs =
    match xs with
    | [] -> []
    | _ -> 
        let x, xs' = select xs in
        x :: (select_sort xs')
```

```ocaml
select_sort [1; 5; 0; 7; 1; 4; 1; 0]
```

```ocaml
let rec split xs =
    match xs with
    | [] -> ([], [])
    | x :: xs' ->
        let l, r = split xs' in
        if (List.length l) > (List.length r) then
            l, (x :: r)
        else
            (x :: l), r
```

```ocaml
split [8; 2; 4; 7; 4; 2; 1]
```

```ocaml
let rec merge xs ys =
    match xs, ys with
    | ([], _) -> ys
    | (_, []) -> xs
    | (x :: xs', y :: ys') ->
        if x < y then
            x :: (merge xs' ys)
        else
            y :: (merge xs ys')
```

```ocaml
merge [1; 4; 4; 8] [2; 2; 7]
```

```ocaml
let rec merge_sort xs =
    if List.length xs <= 1 then
        xs
    else
        let l, r = split xs in
        merge (merge_sort l) (merge_sort r)
```

```ocaml
merge_sort [2; 3; 5; 1; 2; 5; 3; 2; 1; 7]
```
