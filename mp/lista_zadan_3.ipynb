### Zadanie 1

```ocaml
let product xs =
  let rec it xs acc =
    match xs with
    | [] -> acc
    | x :: xs' -> it xs' (acc * x)
  in it xs 1
```

```ocaml
product [1; 2; 3; 4; 2]
```

### Zadanie 3

```ocaml
let build_list n f =
    let rec it n acc =
        match n with
        | -1 -> acc
        | _ -> it (n - 1) (f n :: acc)
    in it (n - 1) []
```

```ocaml
build_list 10 inc
```

```ocaml
let negatives n = build_list n (fun x -> -(x + 1))
```

```ocaml
negatives 5
```

```ocaml
let reciprocals n = build_list n (fun x -> 1. /. float_of_int (x + 1))
```

### Zadanie 6

```ocaml
type 'a tree = Leaf | Node of 'a tree * 'a * 'a tree
```

```ocaml
let rec fold_tree f a t =
  match t with
  | Leaf -> a
  | Node (l, v, r) -> f (fold_tree f a l) v (fold_tree f a r)
```

```ocaml
let tree_product t = fold_tree (fun l v r -> l * v * r) 1 t
```

```ocaml
let t = Node (Node (Leaf, 2, Leaf), 5, Node (Node (Leaf, 6, Leaf), 8, Node (Leaf, 9, Leaf)))
```

```ocaml
tree_product t = 2 * 5 * 6 * 8 * 9
```

```ocaml
let tree_flip t = fold_tree (fun l v r -> Node (r, v, l)) Leaf t
```

```ocaml
tree_flip t
```

```ocaml
let tree_height t = fold_tree (fun l v r -> (max l r) + 1) 0 t
```

```ocaml
tree_height t
```

```ocaml
let tree_span t = (
    fold_tree (fun l v r -> if l = max_int then v else l) max_int t,
    fold_tree (fun l v r -> if r = min_int then v else r) min_int t
)
```

```ocaml
tree_span t
```

```ocaml
let preorder t = fold_tree (fun l v r -> v :: (l @ r)) [] t
```

```ocaml
preorder t
```

### Zadanie 7

```ocaml
type 'a tree = Leaf | Node of 'a tree * 'a * 'a tree
```

```ocaml
let rec flat_append t xs =
    match t with
    | Leaf -> xs
    | Node (l, v, r) ->
        let xs' = flat_append r xs in
        flat_append l (v :: xs')
```

```ocaml
let t = Node (Node (Leaf, 2, Leaf), 5, Node (Node (Leaf, 6, Leaf), 8, Node (Leaf, 9, Leaf)))
```

```ocaml
flat_append t []
```

```ocaml
let flatten t = flat_append t []
```

```ocaml
let tree_sort xs =
    let rec build acc = function
        | [] -> acc
        | x :: xs' -> build (insert_bst x acc) xs'
    in
    let t = build Leaf xs in
    let rec inorder acc = function
        | Leaf -> acc
        | Node (l, v, r) ->
            inorder (v :: (inorder acc r)) l
    in inorder [] t
```

```ocaml
tree_sort [1; 5; 2; 3; 7; 5; 5; 1]
```

```ocaml
type 'a tree = Leaf | Node of 'a tree * 'a * 'a tree
```

```ocaml
let rec first_bigger t =
    match t with
    | Leaf -> assert false
    | Node (Leaf, x, _) -> x
    | Node (l, _, _) -> first_bigger l
```

```ocaml
first_bigger t
```

```ocaml
let rec delete v t =
    match t with
    | Leaf -> assert false
    | Node (l, w, r) ->
        if v < w then
            Node (delete v l, w, r)
        else
        if v > w then
            Node (l, w, delete v r)
        else
            match l, r with
                | Leaf, _ -> r
                | _, Leaf -> l
                | _, _ ->
                    let z = first_bigger r in
                    Node(l, z, delete z r)
```
